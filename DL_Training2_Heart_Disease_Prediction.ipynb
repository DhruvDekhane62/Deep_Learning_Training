 "Heart Disease Prediction"

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
data=pd.read_csv("/content/heart - Copy.csv")
```

```python
data.head()
```

```python
data["target"].unique()
```

```python
data["target"].value_counts()
```

```python
data["target"].value_counts()
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.duplicated().sum()
```

```python
data.isnull().sum()
```

```python
#Imputation
data["trestbps"].fillna(data["trestbps"].mean(),inplace=True)
data["thalach"].fillna(data["thalach"].mean(),inplace=True)
```

```python
data.isnull().sum()
```

```python
#Extraction of Independent variable & Dependent variable usling (.iloc)

X=data.iloc[:,0:-1]
Y=data.iloc[:, -1:]
```

```python
#Data Splitting

from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test=train_test_split(X,Y,test_size=0.2)
```

```python
#Model Creation by Logistic Regression

from sklearn.linear_model import LogisticRegression
lrc=LogisticRegression()
lrc.fit(X_train,Y_train)
```

```python
#Prediction or Testing

Y_pred=lrc.predict(X_test)
```

```python
#Evaluate the performance of metrics

from sklearn.metrics import confusion_matrix
cm=confusion_matrix(Y_test,Y_pred)
print(cm)
```

```python
#Accuracy

from sklearn.metrics import accuracy_score
ac=accuracy_score(Y_test,Y_pred)*100
print(ac)
```

```python
#Precision

from sklearn.metrics import precision_score
ps=precision_score(Y_test,Y_pred)*100
print(ps)
```

```python
#Recall

from sklearn.metrics import recall_score
rs=recall_score(Y_test,Y_pred)*100
print(rs)
```

```python
#F1 Score

from sklearn.metrics import f1_score
f1=f1_score(Y_test,Y_pred)*100
print(f1)
```

```python
from sklearn.metrics import classification_report
print(classification_report(Y_test,Y_pred))
```

# Implementation of KNN

```python
#Data Test Splitting

from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test=train_test_split(X,Y,test_size=0.2)
```

```python
#Create KNN Model

from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier()
knn.fit(X_train,Y_train)
```

```python
#Prediction

Y_pred=knn.predict(X_test)
```

```python
#Confusion Matrix

from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

print(confusion_matrix(Y_test,Y_pred))
print(accuracy_score(Y_test,Y_pred)*100)
print(precision_score(Y_test,Y_pred)*100)
print(recall_score(Y_test,Y_pred)*100)
print(f1_score(Y_test,Y_pred)*100)
```

# Hyperparameter Tunning & Cross Valdation

   * Hyperparameter Tunning is a process of selecting the best or optimal parameter for the model

```python
#GridSearchCV

from sklearn.model_selection import GridSearchCV
from sklearn.neighbors import KNeighborsClassifier

knnc=KNeighborsClassifier()
grid_param={"n_neighbors":[3,5,7,9],"p":[1,2]}

grid_model=GridSearchCV(knnc, grid_param, cv=5)
grid_model.fit(X_train, Y_train.values.ravel())
```

```python
# get the best paramter of the model
grid_model.best_params_
```

```python
final_model=grid_model.best_estimator_
```

```python
Y_pred=final_model.predict(X_test)
```

```python
#Confusion Matrix

from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

print(confusion_matrix(Y_test, Y_pred))
print(accuracy_score(Y_test, Y_pred)*100)
print(precision_score(Y_test, Y_pred)*100)
print(recall_score(Y_test, Y_pred)*100)
print(f1_score(Y_test, Y_pred)*100)
```
