# K MEANS CLUSTERING

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
data=pd.read_csv("/content/cluster_sample.csv")
```

```python
data
```

```python
data.duplicated().sum()
```

```python
data.drop(columns=['Unnamed: 2'],inplace=True)
```

```python
data
```

```python
plt.scatter(data["X1"], data["X2"])
plt.show()
```

```python
from sklearn.cluster import KMeans
kmeans_model = KMeans(n_clusters=5)
kmeans_model.fit(data)
```

```python
#Way 1 for prediction
kmeans_model.labels_
```

```python
#Way 2 for prediction
cluster_labels = kmeans_model.predict(data)
```

```python
data
```

```python
result_data=data.copy()
```

```python
result_data["Cluster_Lables"]=cluster_labels
result_data
```

```python
#Visualization of Clustering

plt.scatter(data["X1"], data["X2"], c=cluster_labels)
plt.show()
```
